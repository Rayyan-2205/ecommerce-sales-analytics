# Data Cleaning

```python
order_df.head(1)
```

```python
def converting_dates(df):
    '''
    converting the date columns
    '''

    date_columns = [
        'order_purchase_timestamp',
        'order_approved_at',
        'order_delivered_carrier_date',
        'order_delivered_customer_date',
        'order_estimated_delivery_date',
        'shipping_limit_date'
    ]

    for col in date_columns:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')

    return df


order_df = converting_dates(order_df)
products_df = converting_dates(products_df)
items_df = converting_dates(items_df)
customers_df = converting_dates(customers_df)
```

```python
def cleanin_report(df):
    print('=' * 40)
    print('            Cleaning Reoprt ')
    print('=' * 40)

    # ================================
    #            Shape
    # ================================

    print('\nShape:')
    print(f'Shape before the Cleaaning: {df.shape}')

    # ================================
    #          Duplicated
    # ================================

    print('\nDuplicated values:')
    print(f'Duplicated values: {df.duplicated().sum()}')

    # ================================
    #         Missing values
    # ================================

    print(f'\nMissing Values: \n{df.isna().sum()}')


    # ================================
    #         Data types
    # ================================

    print(f'\nData types: \n{df.dtypes}')



```

```python
cleanin_report(order_df)
```

```python
cleanin_report(products_df)
```

```python
cleanin_report(items_df)
```

```python
cleanin_report(customers_df)
```

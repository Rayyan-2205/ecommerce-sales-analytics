```python
def converting_dates(df):
    '''
    converting the date columns
    '''

    date_columns = [
        'order_purchase_timestamp',
        'order_approved_at',
        'order_delivered_carrier_date',
        'order_delivered_customer_date',
        'order_estimated_delivery_date',
        'shipping_limit_date'
    ]

    for col in date_columns:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')

    return df


order_df = converting_dates(order_df)
products_df = converting_dates(products_df)
items_df = converting_dates(items_df)
customers_df = converting_dates(customers_df)
```

# Revenue Analysis

### Revenue metrics

```python
def total_price_revenue(df):
    return df['price'].sum()

total_price_revenue(master_df)
```

```python
def total_frieght(df):
    return df['freight_value'].sum()

total_frieght(master_df)
```

```python
def total_revenue(df):
    return total_price_revenue(df) + total_frieght(df)

total_revenue(master_df)
```

```python
def average_order_value(df):
    return total_revenue(df) / total_orders(df)

average_order_value(master_df)
```

```python
def average_price(df):
    return df['price'].mean()

average_price(master_df)
```

```python
def average_frieght(df):
    return df['freight_value'].mean()

average_frieght(master_df)
```

### Revenue by Geography

```python
def revenue_by_state(df):
    return df.groupby('customer_state')['price'].sum().sort_values(ascending=False).head(10)

revenue_by_state(master_df)
```

```python
def revenue_by_city(df):
    return df.groupby('customer_city')['price'].sum().sort_values(ascending=False).head(10)

revenue_by_city(master_df)
```

### Revenue by Category

```python
def revenue_by_category(df):
    return df.groupby('product_category_name')['price'].sum().sort_values(ascending=False).head(10)

revenue_by_category(master_df)
```

```python
def revenue_analysis(df):
    print('=' * 50)
    print('                Revenue Analysis ')
    print('=' * 50)

    print(f'\nTotal Price Revenue: {total_price_revenue(master_df)}')
    print(f'Total Freight: {total_frieght(master_df)}')
    print(f'Total Revenue: {total_revenue(master_df)}')
    print(f'Average Order Value: {average_order_value(master_df)}')
    print(f'Average Price: {average_price(master_df)}')
    print(f'\nAverage Freight: \n{average_frieght(master_df)}')

    print('\nRevenue by Geography')
    print(f'\nTop 10 Sates Revenue: \n{revenue_by_state(master_df)}')
    print(f'\nTop 10 Cities Revenue: \n{revenue_by_city(master_df)}')

    print('\nRevenue by Category')
    print(f'\nTop 10 Category Revenue: \n{revenue_by_category(master_df)}')

    print('=' * 50)

revenue_analysis(master_df)
```

# Customer Analysis

```python
temp_df = master_df.groupby('customer_unique_id')['order_id'].nunique().reset_index(name='order_count')
```

```python
def returning_customers(df):
    return df[df['order_count'] > 1].sort_values(by='order_count',ascending=False).count()

returning_customers(temp_df)
```

```python
def one_time_customer(df):
    return df[df['order_count'] == 1].sort_values(by='order_count',ascending=False).count()

one_time_customer(temp_df)
```

```python
def average_orders_per_customer(df):
    return df['order_count'].sum() / df['customer_unique_id'].count()  

average_orders_per_customer(temp_df)
```

# Order and Time Analysis

```python
master_df['delivery_day'] = (master_df['order_delivered_customer_date'] - master_df['order_purchase_timestamp']).dt.days
```

```python
master_df['month'] = master_df['order_purchase_timestamp'].dt.month_name()
master_df['month_no'] = master_df['order_purchase_timestamp'].dt.month
```

```python
def orders_by_month(df):
    return df.groupby(['month_no', 'month'])['order_id'].nunique().sort_index(level=0)

orders_by_month(master_df)
```

```python
def revenue_by_month(df):
    return df.assign(total = df['price'] + df['freight_value']).groupby(['month_no', 'month'])['total'].sum().sort_index(level=0).reset_index()

revenue_by_month(master_df)
```

```python
master_df['day'] = master_df['order_purchase_timestamp'].dt.day_name()
master_df['day_no'] = master_df['order_purchase_timestamp'].dt.dayofweek
master_df.head(1)
```

```python
def orders_by_day(df):
    return df.groupby(['day_no', 'day'])['order_id'].nunique().sort_index(level=0).reset_index()

orders_by_day(master_df)
```

```python
master_df['hour'] = master_df['order_purchase_timestamp'].dt.hour
```

```python
def orders_by_hour(df):
    return df.groupby('hour')['order_id'].nunique().sort_index(level=0).reset_index()

orders_by_hour(master_df)
```

```python
master_df['year'] = master_df['order_purchase_timestamp'].dt.year
```

```python
def orders_by_year(df):
    return df.groupby('year')['order_id'].nunique().sort_index(level=0).reset_index()

orders_by_year(master_df)
```

```python
def order_status_distribution(df):
    return (df.groupby('order_status')['order_id'].nunique() / df['order_id'].nunique() * 100).round(2).reset_index(name='percentage')

order_status_distribution(master_df)
```

```python
def order_analysis(df):
    print('=' * 50)
    print('                Order Analysis ')
    print('=' * 50)


    print(f'\nOrders by Month: \n{orders_by_month(master_df)}')
    print(f'\nRevenue by Month: \n{revenue_by_month(master_df)}')
    print(f'\nOrders by Day: \n{orders_by_day(master_df)}')
    print(f'\nOrders by Hour: \n{orders_by_hour(master_df)}')
    print(f'\nOrders by Year: \n{orders_by_year(master_df)}')

    print(f'\nOrder Status Distribution: \n{order_status_distribution(master_df)}')

    print('=' * 50)

order_analysis(master_df)
```

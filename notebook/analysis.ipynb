```python
def converting_dates(df):
    '''
    converting the date columns
    '''

    date_columns = [
        'order_purchase_timestamp',
        'order_approved_at',
        'order_delivered_carrier_date',
        'order_delivered_customer_date',
        'order_estimated_delivery_date',
        'shipping_limit_date'
    ]

    for col in date_columns:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')

    return df


order_df = converting_dates(order_df)
products_df = converting_dates(products_df)
items_df = converting_dates(items_df)
customers_df = converting_dates(customers_df)
```

# Business Analysis

### Business Overview

```python
master_df.head()
```

```python
def total_orders(df):
    return df['order_id'].nunique()

(total_orders(master_df))
```

```python
def total_customer(df):
    return df['customer_unique_id'].nunique()

total_customer(master_df)
```

```python
def total_products(df):
    return df['product_id'].count()

total_products(master_df)
```

```python
def unique_products(df):
    return df['product_id'].nunique()

unique_products(master_df)
```

```python
def total_category(df):
    return df['product_category_name'].nunique()

total_category(master_df)
```

```python
def top_states(df):
    return df.groupby('customer_state')['order_id'].nunique().sort_values(ascending=False).head()

top_states(master_df)
```

```python
def top_citites(df):
    return df.groupby('customer_city')['order_id'].nunique().sort_values(ascending=False).head(10)

top_citites(master_df)
```

```python
def business_overview(df):
    print('=' * 50)
    print('                Business Overview ')
    print('=' * 50)

    print(f'\nTotal orders: {total_orders(df)}')
    print(f'Total Customers: {total_customer(df)}')
    print(f'Total Products: {total_products(df)}')
    print(f'Total Unique Products: {unique_products(df)}')
    print(f'Total Product Categories: {total_category(df)}')
    print(f'\nTop 10 Cities by Orders: \n{top_citites(df)}')
    print(f'\nTop 5 States by Orders: \n{top_states(df)}')

    print('=' * 50)
```

```python
business_overview(master_df)
```

# Revenue Analysis

### Revenue metrics

```python
def total_price_revenue(df):
    return df['price'].sum()

total_price_revenue(master_df)
```

```python
def total_frieght(df):
    return df['freight_value'].sum()

total_frieght(master_df)
```

```python
def total_revenue(df):
    return total_price_revenue(df) + total_frieght(df)

total_revenue(master_df)
```

```python
def average_order_value(df):
    return total_revenue(df) / total_orders(df)

average_order_value(master_df)
```

```python
def average_price(df):
    return df['price'].mean()

average_price(master_df)
```

```python
def average_frieght(df):
    return df['freight_value'].mean()

average_frieght(master_df)
```

### Revenue by Geography

```python
def revenue_by_state(df):
    return df.groupby('customer_state')['price'].sum().sort_values(ascending=False).head(10)

revenue_by_state(master_df)
```

```python
def revenue_by_city(df):
    return df.groupby('customer_city')['price'].sum().sort_values(ascending=False).head(10)

revenue_by_city(master_df)
```

### Revenue by Category

```python
def revenue_by_category(df):
    return df.groupby('product_category_name')['price'].sum().sort_values(ascending=False).head(10)

revenue_by_category(master_df)
```

```python
def revenue_analysis(df):
    print('=' * 50)
    print('                Revenue Analysis ')
    print('=' * 50)

    print(f'\nTotal Price Revenue: {total_price_revenue(master_df)}')
    print(f'Total Freight: {total_frieght(master_df)}')
    print(f'Total Revenue: {total_revenue(master_df)}')
    print(f'Average Order Value: {average_order_value(master_df)}')
    print(f'Average Price: {average_price(master_df)}')
    print(f'\nAverage Freight: \n{average_frieght(master_df)}')

    print('\nRevenue by Geography')
    print(f'\nTop 10 Sates Revenue: \n{revenue_by_state(master_df)}')
    print(f'\nTop 10 Cities Revenue: \n{revenue_by_city(master_df)}')

    print('\nRevenue by Category')
    print(f'\nTop 10 Category Revenue: \n{revenue_by_category(master_df)}')

    print('=' * 50)

revenue_analysis(master_df)
```

# Product Analysis

```python
def best_selling_product(df):
    return df['product_id'].value_counts().head(10)

best_selling_product(master_df)
```

```python
def highest_revenue_product(df):
    return df.groupby('product_id')['price'].sum().sort_values(ascending=False).head(10)

highest_revenue_product(master_df)
```

```python
def best_selling_categories(df):
    return df.groupby('product_category_name')['order_id'].count().sort_values(ascending=False).head(10)

best_selling_categories(master_df)
```

```python
def highest_revenue_category(df):
    return df.groupby('product_category_name')['price'].sum().sort_values(ascending=False).head(10)

highest_revenue_category(master_df)
```

```python
def average_product_price_by_category(df):
    return df.groupby('product_category_name')['price'].mean().sort_values(ascending=False).head(10)

average_product_price_by_category(master_df)
```

### Product Characteristics

```python
def average_weight_by_category(df):
    return df.groupby('product_category_name')['product_weight_g'].mean().head(10)

average_weight_by_category(master_df)
```

```python
def average_product_dimensions(df):
        return df.groupby('product_category_name')[['product_length_cm', 'product_height_cm', 'product_width_cm']].mean().head(10)

average_product_dimensions(master_df)
```

```python
def heaviest_product(df):
    return df.groupby('product_category_name')['product_weight_g'].max().sort_values(ascending=False).head(10)

heaviest_product(master_df)
```

```python
def longest_description(df):
    return df.groupby('product_category_name')['product_description_lenght'].max().sort_values(ascending=False).head(10)

longest_description(master_df)
```

```python
def product_analysis(df):
    print('=' * 50)
    print('                Product Analysis ')
    print('=' * 50)

    print('\nProduct Performance')

    print(f'\nBest Selling Products: \n{best_selling_product(master_df)}')
    print(f'\nHighest Revenue Products: \n{highest_revenue_product(master_df)}')
    print(f'\nBest Selling category: \n{best_selling_categories(master_df)}')
    print(f'\nHighest Revenue Category: \n{highest_revenue_category(master_df)}')
    print(f'\nAverage Product price by Category: \n{average_product_price_by_category(master_df)}')

    print('\nProduct Characteristics')
    print(f'\nAverage weight by Category: \n{average_weight_by_category(master_df)}')
    print(f'\nAverage product dimensions by Category: \n{average_product_dimensions(master_df)}')
    print(f'\nHeaviest Products by Category: \n{heaviest_product(master_df)}')
    print(f'\nLongest product description by Category: \n{longest_description(master_df)}')

    print('=' * 50)

product_analysis(master_df)
```

# Customer Analysis

### Customer Overview

```python
master_df.head(1)
```

```python
temp_df = master_df.groupby('customer_unique_id')['order_id'].nunique().reset_index(name='order_count')
```

```python
def returning_customers(df):
    return df[df['order_count'] > 1].sort_values(by='order_count',ascending=False).count()

returning_customers(temp_df)
```

```python
def one_time_customer(df):
    return df[df['order_count'] == 1].sort_values(by='order_count',ascending=False).count()

one_time_customer(temp_df)
```

```python
def average_orders_per_customer(df):
    return df['order_count'].sum() / df['customer_unique_id'].count()  

average_orders_per_customer(temp_df)
```

### Customer Geography

```python
def customers_by_state(df):
    return df.groupby('customer_state')['customer_unique_id'].nunique().sort_values(ascending=False).head(10)

customers_by_state(master_df)
```

```python
def customers_by_city(df):
    return df.groupby('customer_city')['customer_unique_id'].nunique().sort_values(ascending=False).head(10)

customers_by_city(master_df)
```

```python
def average_customer_spending(df):
    return total_revenue(df).sum() / df['customer_unique_id'].nunique()

average_customer_spending(master_df)
```

```python
def top_customer_state_by_spending(df):
    return  df.assign(total = df['price'] + df['freight_value']).groupby('customer_state')['total'].mean().sort_values(ascending=False).head(10)

top_customer_state_by_spending(master_df)
```
